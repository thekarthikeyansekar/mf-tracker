```python
mapper = {
    "DSP Mid Cap Fund" : "DSP Midcap Fund",
    "DSP Multi-Asset Allocation Fund" : "DSP Multi Asset Allocation Fund",
    "DSP Silver ETF FoF" : "DSP Silver ETF Fund of Fund",
    "Axis ELSS Tax Saver Fund" : "Axis ELSS- Tax Saver Fund",
    "Axis MultiCap Fund" : "Axis Multicap Fund",
    "Axis Retirement Savings Fund - Aggressive Plan" : "Axis Retirement Fund - Aggressive Plan"
}
```

```python
scheme_code_map = dict()
```

```python
for f in funds:
    try:
        f_alt = mapper.get(f,f)
        if f in mf_prices_df['Scheme Name'].values:
            tmp_code = int(mf_prices_df[mf_prices_df['Scheme Name'] == f].iloc[0]['Scheme Code'])
        elif f_alt in mf_prices_df['Scheme Name'].values:
            tmp_code = int(mf_prices_df[mf_prices_df['Scheme Name'] == f_alt].iloc[0]['Scheme Code'])
        
        scheme_code_map[f] = tmp_code
        del tmp_code
    except Exception as e:
        
        print(f"{f} - fund not mapped")
```
